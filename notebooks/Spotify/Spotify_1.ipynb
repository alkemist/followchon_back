```python
from IPython.display import display

import sys
import os
import pandas as pd

sys.path.append(os.path.abspath("./"))
import helpers as _
```

```python
data_dir = 'Spotify/data'

json_files = _.list_files(data_dir, r'.*\.(json)$')
json_files
```

```python
df = pd.DataFrame()

for json_file in json_files:
    df_file = pd.read_json(f'{data_dir}/{json_file}')
    df = pd.concat([df, df_file])

df.to_csv(f'{data_dir}/Streaming_History_Audio.csv')

df
```

```python
df = pd.read_csv(f'{data_dir}/Streaming_History_Audio.csv')

df
```

```python
df_filtered = df[['ts', 
    'master_metadata_track_name', 'master_metadata_album_artist_name', 'master_metadata_album_album_name', 
    'spotify_track_uri', 'reason_start', 'reason_end', 'skipped', 'ms_played']]

df_filtered['datetime'] = pd.to_datetime(df['ts'])
df_filtered['uid'] = df_filtered['spotify_track_uri'].str.replace('spotify:track:', '')

df_filtered.to_csv(f'{data_dir}/Streaming_History_filtered.csv')
```

```python
df_filtered['spotify_track_uri'].unique().shape[0]


df_songs = df_filtered[['master_metadata_track_name', 'master_metadata_album_artist_name', 'master_metadata_album_album_name', 
    'uid']].drop_duplicates()

df_songs = df_songs.rename(columns={'master_metadata_track_name': 'track', 'master_metadata_album_artist_name': 'artist', 'master_metadata_album_album_name': 'album'})

df_songs[['uid', 'artist', 'album', 'track']].to_csv(f'{data_dir}/Streaming_tracks.csv', index=False)
```
